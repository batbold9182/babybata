```python
%%writefile train.py
# Fine-tune GPT-2 into BabyBata.
# Run:  python train.py        (works on a laptop CPU, much faster on a GPU / Google Colab)
#
# It's the same loop as your from-scratch network:
#   forward -> loss -> backward -> step
# except the starting weights come from GPT-2 instead of random numbers.

import os
import random
import time

import torch
from transformers import GPT2LMHeadModel, GPT2TokenizerFast

HERE = os.path.dirname(os.path.abspath(__file__))
DATA_FILE = os.path.join(HERE, "data", "babybata.txt")        # hand-written + teach mode
GENERATED_FILE = os.path.join(HERE, "data", "generated.txt")  # made by augment.py
OUT_DIR = os.path.join(HERE, "babybata-model")

# Where training STARTS from. Plain GPT-2 = a clean start, no old habits
# (v1 had memorized a wrong email). To continue from a saved version instead,
# point this at its folder, e.g. os.path.join(HERE, "babybata-model-v1").
BASE_MODEL = "gpt2"
EPOCHS = 6               # best validation was at epoch 4; later epochs only overfit
FREEZE_EMBEDDINGS = True  # don't train the 38.6M-weight vocabulary table: ~faster, less memory
LEARNING_RATE = 5e-5     # tiny, so we adjust GPT-2's knowledge instead of overwriting it
BATCH_SIZE = 4
MAX_TOKENS = 128
VALIDATION_FRACTION = 0.15
SEED = 42

random.seed(SEED)
torch.manual_seed(SEED)
device = "cuda" if torch.cuda.is_available() else "cpu"


# ---------------------------------------------------------------
# 1. DATA: read User/BabyBata pairs from the text file
# ---------------------------------------------------------------
def load_pairs(path):
    pairs = []
    with open(path, encoding="utf-8") as f:
        blocks = f.read().split("\n\n")
    for block in blocks:
        lines = [l for l in block.strip().splitlines() if l and not l.startswith("#")]
        if len(lines) == 2 and lines[0].startswith("User:") and lines[1].startswith("BabyBata:"):
            question = lines[0][len("User:"):].strip()
            answer = lines[1][len("BabyBata:"):].strip()
            pairs.append((question, answer))
    return pairs


def prompt_for(question):
    """The exact format the model sees. chat.py must use the same one."""
    return f"User: {question}\nBabyBata:"


# ---------------------------------------------------------------
# 2. TOKENIZE, with loss masking:
#    the model is only graded on BabyBata's answer, not on the question.
#    Label -100 means "ignore this position in the loss".
# ---------------------------------------------------------------
def encode(pair, tokenizer):
    question, answer = pair
    prompt_ids = tokenizer.encode(prompt_for(question))
    answer_ids = tokenizer.encode(" " + answer) + [tokenizer.eos_token_id]  # eos = "stop talking"
    ids = (prompt_ids + answer_ids)[:MAX_TOKENS]
    labels = ([-100] * len(prompt_ids) + answer_ids)[:MAX_TOKENS]
    return ids, labels


def make_batch(examples, pad_id):
    """Pad every example in the batch to the same length."""
    longest = max(len(ids) for ids, _ in examples)
    input_ids, labels, attention = [], [], []
    for ids, labs in examples:
        pad = longest - len(ids)
        input_ids.append(ids + [pad_id] * pad)
        labels.append(labs + [-100] * pad)         # padding is never graded
        attention.append([1] * len(ids) + [0] * pad)  # and never looked at
    return (torch.tensor(input_ids, device=device),
            torch.tensor(labels, device=device),
            torch.tensor(attention, device=device))


def batches(examples, pad_id, shuffle):
    order = list(range(len(examples)))
    if shuffle:
        random.shuffle(order)
    for s in range(0, len(order), BATCH_SIZE):
        yield make_batch([examples[i] for i in order[s:s + BATCH_SIZE]], pad_id)


def validation_loss(model, examples, pad_id):
    model.eval()
    total, count = 0.0, 0
    with torch.no_grad():
        for ids, labels, attn in batches(examples, pad_id, shuffle=False):
            total += model(input_ids=ids, attention_mask=attn, labels=labels).loss.item() * len(ids)
            count += len(ids)
    model.train()
    return total / count


def prepare_data(tokenizer):
    """Same train/validation split every time (compress.py relies on this)."""
    pairs = load_pairs(DATA_FILE)
    if os.path.exists(GENERATED_FILE):
        pairs += load_pairs(GENERATED_FILE)
    random.Random(SEED).shuffle(pairs)
    n_val = max(1, int(len(pairs) * VALIDATION_FRACTION))
    val_pairs, train_pairs = pairs[:n_val], pairs[n_val:]
    train = [encode(p, tokenizer) for p in train_pairs]
    val = [encode(p, tokenizer) for p in val_pairs]
    pad_id = tokenizer.eos_token_id  # GPT-2 has no pad token; masked out anyway
    return train, val, pad_id


def fine_tune(model, tokenizer, train, val, pad_id, epochs, learning_rate, out_dir):
    """The training loop. Saves to out_dir whenever validation loss improves."""
    # AdamW = gradient descent with momentum and per-weight step sizes (lesson 9 material)
    # Only weights that are being trained (not frozen) go to the optimizer.
    optimizer = torch.optim.AdamW([p for p in model.parameters() if p.requires_grad], lr=learning_rate)

    best = validation_loss(model, val, pad_id)
    print(f"before training: validation loss = {best:.3f}")

    for epoch in range(1, epochs + 1):
        start, train_total, seen = time.time(), 0.0, 0
        for ids, labels, attn in batches(train, pad_id, shuffle=True):
            loss = model(input_ids=ids, attention_mask=attn, labels=labels).loss  # forward + loss
            loss.backward()          # backward: PyTorch does the chain rule for you
            optimizer.step()         # step: move every weight downhill
            optimizer.zero_grad()    # reset gradients for the next batch
            train_total += loss.item() * len(ids)
            seen += len(ids)

        val_loss = validation_loss(model, val, pad_id)
        note = ""
        if val_loss < best:
            best = val_loss
            model.save_pretrained(out_dir)
            tokenizer.save_pretrained(out_dir)
            note = "  <- best so far, saved"
        print(f"epoch {epoch}: train loss = {train_total / seen:.3f}   "
              f"validation loss = {val_loss:.3f}   ({time.time() - start:.0f}s){note}")
    return best


def main():
    print(f"device: {device}")
    tokenizer = GPT2TokenizerFast.from_pretrained(BASE_MODEL)
    model = GPT2LMHeadModel.from_pretrained(BASE_MODEL).to(device)
    print(f"loaded {BASE_MODEL}: {sum(p.numel() for p in model.parameters()):,} parameters")

    if FREEZE_EMBEDDINGS:
        # The word table (also used as the output layer) and position table stay as they are.
        model.transformer.wte.weight.requires_grad_(False)
        model.transformer.wpe.weight.requires_grad_(False)
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"training {trainable:,} of them (the rest are frozen)")

    train, val, pad_id = prepare_data(tokenizer)
    print(f"{len(train)} training examples, {len(val)} validation examples\n")

    fine_tune(model, tokenizer, train, val, pad_id, EPOCHS, LEARNING_RATE, OUT_DIR)

    if not os.path.exists(OUT_DIR):
        print("\nValidation never improved, so nothing was saved. Try more data or more epochs.")
    else:
        print(f"\nBabyBata saved to {OUT_DIR}.  Talk to it:  python chat.py")


if __name__ == "__main__":
    main()
```

```python
%%writefile facts.py
# Exact facts BabyBata must never get wrong.
#
# A language model stores word PATTERNS, not text, so it can blend similar
# strings (it once turned the email into "batbold9182745" by mixing in the
# GitHub handle). So in the training data we write a slot like {EMAIL}
# instead, the model learns WHEN to say it, and fill() swaps in the real value.
#
# Change a value here and every answer updates, no retraining needed.

FACTS = {
    "EMAIL": "batbold918273645@gmail.com",
    "GITHUB": "github.com/batbold9182",
    "LEETCODE": "leetcode.com/u/batbold9182",
    "FULL_NAME": "Batbold Samdan",
    "UNIVERSITY": "Vizja University Warsaw",
    "GPA": "4.0 out of 5",
}


def fill(text):
    """Model output -> what the visitor sees:  '{EMAIL}' -> 'batbold918273645@gmail.com'"""
    for slot, value in FACTS.items():
        text = text.replace("{" + slot + "}", value)
    return text


def to_slots(text):
    """The reverse, for saving training data:  'batbold918273645@gmail.com' -> '{EMAIL}'"""
    for slot, value in sorted(FACTS.items(), key=lambda kv: -len(kv[1])):  # longest first
        text = text.replace(value, "{" + slot + "}")
    return text
```

```python
%%writefile check.py
# Fact-check BabyBata: does it get the important things right?
# Run:  python check.py                      <- babybata-model/
#       python check.py babybata-model-v1    <- any other saved version, to compare
#
# The loss number never warned us the email was wrong. This does.
# Answers are generated greedily (no randomness), so results are repeatable.

import sys

import torch
from transformers import GPT2LMHeadModel, GPT2TokenizerFast
from transformers.utils import logging

from facts import fill
from train import OUT_DIR, prompt_for

logging.set_verbosity_error()

# (question, text the answer MUST contain). Phrased differently from the training data on purpose.
TESTS = [
    ("who are you", "BabyBata"),
    ("whats your name", "BabyBata"),
    ("how do i contact batbold", "{EMAIL}"),
    ("what's his email address?", "{EMAIL}"),
    ("where is batbold's github", "{GITHUB}"),
    ("where does batbold go to university", "{UNIVERSITY}"),
    ("what's batbold's gpa", "{GPA}"),
    ("what projects has batbold made", "Express"),
    ("tell me about smart campus", "university"),
    ("what's the salon app", "salon"),
    ("is batbold looking for work", "internship"),
    ("where is batbold from", "Mongolia"),
    ("what does batbold do for fun", "basketball"),
    ("what's the weather", "only know about Batbold"),
    ("can you do my homework", "only know about Batbold"),
]


def main():
    path = sys.argv[1] if len(sys.argv) > 1 else OUT_DIR
    tokenizer = GPT2TokenizerFast.from_pretrained(path)
    model = GPT2LMHeadModel.from_pretrained(path).eval()

    passed = 0
    for question, must_contain in TESTS:
        ids = tokenizer.encode(prompt_for(question), return_tensors="pt")
        with torch.no_grad():
            out = model.generate(ids, attention_mask=torch.ones_like(ids), max_new_tokens=60,
                                 do_sample=False, pad_token_id=tokenizer.eos_token_id,
                                 eos_token_id=tokenizer.eos_token_id)
        raw = tokenizer.decode(out[0][ids.shape[1]:], skip_special_tokens=True).split("\nUser:")[0].strip()
        ok = must_contain.lower() in raw.lower()
        passed += ok
        print(f"{'PASS' if ok else 'FAIL'}  {question}\n      -> {fill(raw)}")
    print(f"\n{path}: {passed}/{len(TESTS)} facts correct")


if __name__ == "__main__":
    main()
```

```python
from transformers import GPT2LMHeadModel, GPT2TokenizerFast
from train import OUT_DIR, prompt_for
from facts import fill

device = "cuda" if torch.cuda.is_available() else "cpu"
tokenizer = GPT2TokenizerFast.from_pretrained(OUT_DIR)
babybata = GPT2LMHeadModel.from_pretrained(OUT_DIR).to(device).eval()
plain_gpt2 = None


def reply(model, question):
    ids = tokenizer.encode(prompt_for(question), return_tensors="pt").to(device)
    with torch.no_grad():
        out = model.generate(ids, attention_mask=torch.ones_like(ids), max_new_tokens=80,
                             do_sample=True, temperature=0.3, top_p=0.9, repetition_penalty=1.05,
                             pad_token_id=tokenizer.eos_token_id, eos_token_id=tokenizer.eos_token_id)
    text = tokenizer.decode(out[0][ids.shape[1]:], skip_special_tokens=True)
    return fill(text.split("\nUser:")[0].strip())  # {EMAIL} -> the real email


def ask(question, compare=False):
    global plain_gpt2
    print("You:        ", question)
    print("BabyBata:   ", reply(babybata, question))
    if compare:
        if plain_gpt2 is None:
            plain_gpt2 = GPT2LMHeadModel.from_pretrained("gpt2").to(device).eval()
        print("plain GPT-2:", reply(plain_gpt2, question))
    print()


ask("Who are you?", compare=True)
ask("What kind of projects do you build?", compare=True)
ask("How do I get in touch?")
ask("What's his GitHub?")
ask("What's the best pizza topping?")
```

## 6. Compress BabyBata (optional, ~2 minutes)
Quantizes the weights to 8 and 4 bits (by hand), removes 6 of the 12 layers, and scores every version.
Needs step 3 to have run in this session.

```python
%%writefile compress.py
# Shrink BabyBata and measure what each cut costs.
# Run AFTER train.py (needs babybata-model/):  python compress.py
#
#   1. Quantization (written by hand): store each weight with fewer bits
#   2. Layer removal: keep 6 of GPT-2's 12 transformer layers, then retrain briefly
#   3. Both together
#
# Every version is scored on the same validation set (lower loss = better)
# and answers the same questions, so you can see the trade-off.

import copy
import os

import torch
from transformers import GPT2LMHeadModel, GPT2TokenizerFast

from train import OUT_DIR, device, fine_tune, prepare_data, prompt_for, validation_loss

SMALL_DIR = OUT_DIR + "-small"
KEEP_LAYERS = [0, 2, 4, 6, 8, 10]   # every other layer, out of 0..11
RETRAIN_EPOCHS = 6
RETRAIN_LR = 1e-4                   # a bit higher: the cut model has more to repair
TEST_QUESTIONS = ["Who are you?", "What projects has Batbold built?", "How can I contact Batbold?"]


# ---------------------------------------------------------------
# Quantization by hand
#   For each weight matrix, row by row:
#     scale   = biggest |weight| in the row / largest integer we can store
#     stored  = round(weight / scale)        <- small integer, e.g. -127..127
#     used    = stored * scale               <- what the network computes with
#   Rounding error is the only "damage". We simulate it in float so we can
#   measure quality; a real deployment would store the integers + scales.
# ---------------------------------------------------------------
def quantize(model, bits):
    q = copy.deepcopy(model)
    top = 2 ** (bits - 1) - 1                    # 8 bits -> 127, 4 bits -> 7
    with torch.no_grad():
        for p in q.parameters():
            if p.dim() != 2:                     # biases, LayerNorm: tiny, keep exact
                continue
            scale = p.abs().amax(dim=1, keepdim=True).clamp(min=1e-12) / top
            p.copy_(torch.clamp(torch.round(p / scale), -top - 1, top) * scale)
    return q


def size_mb(model, bits=32):
    """File size if 2D weights use `bits` each (+ one 16-bit scale per row)
    and everything else stays 32-bit."""
    total_bits = 0
    for p in model.parameters():
        if p.dim() == 2 and bits < 32:
            total_bits += p.numel() * bits + p.shape[0] * 16
        else:
            total_bits += p.numel() * 32
    return total_bits / 8 / 1e6


def answer(model, tokenizer, question):
    model.eval()  # validation_loss() leaves the model in training mode (dropout on)
    ids =tokenizer.encode(prompt_for(question), return_tensors="pt").to(device)
    with torch.no_grad():
        out = model.generate(ids, attention_mask=torch.ones_like(ids), max_new_tokens=50,
                             do_sample=False,  # greedy: same input -> same output, fair comparison
                             pad_token_id=tokenizer.eos_token_id, eos_token_id=tokenizer.eos_token_id)
    text = tokenizer.decode(out[0][ids.shape[1]:], skip_special_tokens=True)
    return text.split("\nUser:")[0].strip()


def main():
    if not os.path.exists(OUT_DIR):
        raise SystemExit("No BabyBata yet. Run train.py first.")

    tokenizer = GPT2TokenizerFast.from_pretrained(OUT_DIR)
    full = GPT2LMHeadModel.from_pretrained(OUT_DIR).to(device).eval()
    train, val, pad_id = prepare_data(tokenizer)

    # ---- layer removal: cut, then let the remaining layers adapt ----
    print(f"Removing layers: keeping {len(KEEP_LAYERS)} of {full.config.n_layer} -> {KEEP_LAYERS}")
    small = copy.deepcopy(full)
    small.transformer.h = torch.nn.ModuleList([small.transformer.h[i] for i in KEEP_LAYERS])
    small.config.n_layer = len(KEEP_LAYERS)
    for new_index, block in enumerate(small.transformer.h):  # kept layers were 0,2,4.. -> now 0,1,2..
        if hasattr(block.attn, "layer_idx"):
            block.attn.layer_idx = new_index
    small.train()
    print(f"right after cutting: validation loss = {validation_loss(small, val, pad_id):.3f}  (damaged!)")
    print("retraining the smaller model...")
    fine_tune(small, tokenizer, train, val, pad_id, RETRAIN_EPOCHS, RETRAIN_LR, SMALL_DIR)
    small = GPT2LMHeadModel.from_pretrained(SMALL_DIR).to(device).eval()  # best checkpoint

    variants = [
        ("full, 32-bit (original)", full, 32),
        ("full, 8-bit", quantize(full, 8), 8),
        ("full, 4-bit", quantize(full, 4), 4),
        ("6 layers, 32-bit", small, 32),
        ("6 layers, 8-bit", quantize(small, 8), 8),
        ("6 layers, 4-bit", quantize(small, 4), 4),
    ]

    print("\n" + "=" * 72)
    print(f"{'version':26} {'parameters':>12} {'size':>9} {'val loss':>9}")
    print("-" * 72)
    for name, model, bits in variants:
        params = sum(p.numel() for p in model.parameters())
        loss = validation_loss(model, val, pad_id)
        print(f"{name:26} {params / 1e6:10.1f}M {size_mb(model, bits):7.0f}MB {loss:9.3f}")
    print("=" * 72)

    for question in TEST_QUESTIONS:
        print(f"\nQ: {question}")
        for name, model, _ in variants:
            print(f"  {name:26} {answer(model, tokenizer, question)}")

    print(f"\nSaved the 6-layer model to {SMALL_DIR}")


if __name__ == "__main__":
    main()
```
